```python
import pandas as pd
import oracledb

# --- Database connection details ---
username = "system"
password = "PassWord"
dsn = "localhost:1521/XE"  # Use XEPDB1 if that's where the table is now


# Connect to Oracle
connection = oracledb.connect(user=username, password=password, dsn=dsn)
print("✅ Connected to Oracle Database")
df = pd.read_sql("Select * from emp",con=connection)
print(df)
```

```python
connection = oracledb.connect(user=username, password=password, dsn=dsn)
print("✅ Connected to Oracle Database")
df = pd.read_sql("Select empno,ename,deptno from emp",con=connection)
print(df)
```

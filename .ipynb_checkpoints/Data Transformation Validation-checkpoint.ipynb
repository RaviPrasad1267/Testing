```python
# 🟦 1. What is pd.merge()?
# pd.merge() combines two DataFrames based on a common column (called a key).
# It is similar to SQL joins.
select * from emp join/left join / right join dept on deptno=departmentno
```

```python
# Syntax:
# pd.merge(
#     left=df1,
#     right=df2,
#     how="join_type",
#     on="column_name"
# )
# Parameters:
# left, right → DataFrames to merge
# how → type of join: "inner", "left", "right", "outer"
# on → column name to join on
# left_on, right_on → if column names differ
# indicator=True → shows match/mismatch status
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["Aman", "Bhavya", "Chirag", "Divya"]
})

df2 = pd.DataFrame({
    "id12": [2, 3, 5],
    "salary": [50000, 60000, 45000]
})
```

```python
#  5. Joining on Different Column Names (left_on, right_on)
inner_join_df =  pd.merge(
    left=df1,
    right=df2,
    how="inner",
    on="id"
)
inner_join_df
```

```python
left_join_df =  pd.merge(
    left=df1,
    right=df2,
    how="left",
    on="id"
)
left_join_df
```

```python
right_join_df =  pd.merge(
    left=df1,
    right=df2,
    how="right",
    on="id"
)
right_join_df
```

```python
outer_join_df =  pd.merge(
    left=df1,
    right=df2,
    how="outer",
    on="id"
)
outer_join_df
```

```python
#  method 2
outer_join_df =  df1.merge(
    df2,
    left_on="id",
    right_on="id12",
    how="outer"
    
)
outer_join_df
```

```python
# 🟨 6. Detecting Mismatches → indicator=True
# This helps in validating the join — shows where the row came from:
# left_only → present only in df1
# right_only → present only in df2
# both → successfully matched
```

```python
#  pd.merge(df1, df2, on="id", how="outer", indicator=True)
outer_join_df =  df1.merge(
    df2,
    left_on="id",
    right_on="id12",
    how="outer",
    indicator=True
    
)
outer_join_df
```

```python
# 🟪 7. How to filter unmatched rows
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["A", "B", "C", "D"]
})

df2 = pd.DataFrame({
    "id": [2, 4],
    "age": [30, 40]
})
```

```python
#  ✅ Example 1: Get rows in df1 that DO NOT match df2
left_join_df =  pd.merge(
    left=df1,
    right=df2,
    how="left",
    on="id",
    indicator=True
)
left_join_df
left_only_records=left_join_df[left_join_df["_merge"]=="left_only"]
left_only_records
```

```python
#  ✅ Example 2: Get rows in df2 that DO NOT match df1
#  ✅ Example 1: Get rows in df1 that DO NOT match df2
left_join_df =  pd.merge(
    left=df1,
    right=df2,
    how="right",
    on="id",
    indicator=True
)
left_join_df
left_only_records=left_join_df[left_join_df["_merge"]=="right_only"]
left_only_records
```

```python
#  🟦 1. Pivot → df.pivot()
# Pivot reshapes data by turning unique column values into new columns.
#  df.pivot(index="row_col", columns="column_to_expand", values="value_col")
```

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["A", "A", "B", "B"],
    "year": [2021, 2022, 2021, 2022],
    "sales": [100, 150, 200, 250]
})
df

pivoted = df.pivot(index="name", columns="year", values="sales")
print(pivoted)
```

```python
#  🟩 2. Pivot Table → df.pivot_table()
# pivot_table() is smarter than pivot:
# Allows duplicates
# Allows aggregations (sum, mean…)
```

```python
df = pd.DataFrame({
    "name": ["A", "A", "A", "B", "B"],
    "year": [2021, 2021, 2022, 2021, 2022],
    "sales": [100, 150, 130, 200, 250]
})
df
pivot_tbl = df.pivot_table(
    index="name",
    columns="year",
    values="sales",
    aggfunc="mean"
)

pivot_tbl
```

```python
#  🟥 3. Melt → pd.melt()
# Melt converts wide format → long format.
#  pd.melt(df, id_vars=[...], value_vars=[...], var_name="col", value_name="val")
```

```python
df = pd.DataFrame({
    "name": ["A", "B"],
    "math": [90, 85],
    "science": [88, 92]
})

melted = pd.melt(
    df,
    id_vars=["name"],
    value_vars=["math", "science"],
    var_name="subject",
    value_name="marks"
)

print(melted)
```

```python
#  🟨 5. Map → df['col'].map()
#  Maps values of a column using a dictionary or function.
df = pd.DataFrame({
    "grade": ["A", "B", "C", "A"]
})

mapping = {"A": 90, "B": 80, "C": 70}

df["marks"] = df["grade"].map(mapping)        # Other method df.assign 
print(df)
```

```python
# df.isna() is a pandas function that checks for missing values (also called NaN, None, or NaT).
# It returns a DataFrame of booleans:
# True → the value is missing
# False → the value is NOT missing
# It is identical to df.isnull() (both do the same thing).
```

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["A", "B", None,None],
    "age": [23, None, 30,None]
})

df
```

```python
# ✅ 1. Count missing values Per column:
df.isna().sum()
```

```python
# Total missing value in df
df.isna().sum().sum()       
# sql       <------------------------------
# Pthon or pandas ----------------------------->
```

```python
# ✅ 2. Filter rows with missing values
df[df.isna().any(axis=1)]
```

```python
#  Rows where ALL values are missing:
df[df.isna().all(axis=1)]
```

```python
# ✅ 3. Select rows where a specific column is missing
df[df[["age","name"]].isna()]
```

```python
# null check multiple columns
df[df[["age", "name"]].isna().any(axis=1)]
```

```python
# 8. na_values – treat extra strings as NaN (missing values)
# When reading a CSV, pandas automatically treats some values as NaN (missing), like:
# empty cells
# "NA"
# "NaN"
# "null"
# But real-world CSV files often contain custom missing-value markers, such as:
# "?"
# "N/A"
# "None"
# "missing"
# "—"
# "no data"
# The na_values parameter lets you tell pandas which extra strings should be considered missing (NaN).
# ✔ 1. Single value
# Treat "?" as NaN:
df = pd.read_csv("emp.csv", na_values="?")
# ✔ 2. Multiple missing-value markers
df = pd.read_csv("emp.csv", na_values=["?", "N/A", "none", "-"])  # Now all of them become real NaN.
# When we want to use for different columns then 
df = pd.read_csv(
    "emp.csv",
    na_values={
        "ENAME": ["NoName"],
        "SAL": [-1],
        "DEPTNO": [999]
    }
)
df
```

```python
# ❗ What happens when you set keep_default_na=False?
# You tell pandas:
# “Do NOT treat the usual strings like ‘NA’, ‘N/A’, ‘None’, ‘nan’ as missing.
# Only treat missing values I specify in na_values.”
# So pandas stops its automatic NA detection.
```

```python
# Tom,N/A
# ✔ Case 1 — Default behavior (keep_default_na=True)  # “NA” and “N/A” become NaN automatically.
# ✔ Case 2 — Disable default NA detection # df = pd.read_csv("emp.csv", keep_default_na=False)
# ✔ Case 3 — Combine with custom missing values
df = pd.read_csv(
    "emp.csv",
    keep_default_na=False,
    na_values=["???", "none", "--"]
)
df
```

```python
# 📌 10. nrows – read only X rows from a CSV
# The nrows parameter tells pandas to read only the first N rows of the CSV file.
# It is extremely useful when:
# your file is very large
# you want to preview the data
# you want to test code quickly
# you want to load only a small portion of the dataset
df = pd.read_csv("emp.csv", nrows=5)
df
```

```python
# 📌 11. skiprows – skip rows at the start (or anywhere)
# The skiprows parameter tells pandas:
# “Do not read these rows from the CSV file.”
# You can skip:
# the first few rows
# a specific list of rows
# rows that match a condition (using a function)
# skiprows is extremely useful when your CSV contains:
# metadata or notes at the top
# unnecessary header information
# blank rows
# bad rows
# multiple header lines
```

```python
# ✅ 1. Skip first N rows
# This is the most common usage.
# Example: skip the first 2 rows
df = pd.read_csv("emp.csv", skiprows=2) # This removes rows 0 and 1 (index in the file, not DataFrame index).
df
```

```python
df = pd.read_csv("emp.csv", skiprows=2) # This removes rows 0 and 1 (index in the file, not DataFrame index).
df
```

```python
# ✅ 2. Skip specific rows (using list)
# If you only want to skip certain lines, use a list.
# Example: skip rows 1 and 3
df = pd.read_csv("emp.csv", skiprows=[1, 3])
df
```

```python
# ✅ 3. Skip rows using a function (advanced but powerful)
# You can pass a function that returns True for rows you want to skip.
# Example: skip all rows starting with #
# Comment
# More comments
EMPNO,ENAME,SAL
101,John,3000
df = pd.read_csv("emp.csv", skiprows=lambda x: x < 2) # This skips the first two rows
```

```python
# ✅ 4. Skip a range + read only some rows
# Often used with nrows to read specific blocks from large CSVs.
# Example: read rows 100–199
df = pd.read_csv("emp.csv", skiprows=5, nrows=13)
df
```

```python
# 📌 12. skipfooter – skip rows at the bottom of the file
# skiprows removes rows from the top
# skipfooter removes rows from the bottom
# ❗ But why do we need engine="python"?
# By default, pandas uses the C engine for reading CSV—fast but limited.
# The C engine cannot handle skipfooter. So pandas requires:
df=pd.read_csv("emp.csv", skipfooter=2, engine="python")
df
```

```python
# 📌 13. chunksize – load file in chunks (iterator / generator)
# When you set chunksize, pandas does not load the entire CSV into memory.
# Instead, it reads the file piece by piece (chunk by chunk).
# Each chunk is returned as a DataFrame, and you can process each chunk incrementally.
# This is extremely helpful when working with:
# very large CSV files (GB-sized)
# low RAM systems ,stream processing. batch processing
```

```python
#Example 1 — Read in chunks of 500 rows
for chunk in pd.read_csv("emp.csv", chunksize=5):
    print(chunk)
```

```python
# 📘 Example 2 — Count total number of rows in a huge file
total = 0
for chunk in pd.read_csv("emp.csv", chunksize=1000):
    total += len(chunk)

print("Total rows:", total)
```

```python
# 📘 Example 3 — Process data chunk-by-chunk
total_salary = 0

for chunk in pd.read_csv("emp.csv", chunksize=10000):
    total_salary += chunk['SAL'].sum()

print(total_salary)
```
